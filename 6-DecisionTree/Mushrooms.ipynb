```python
# Decision Tree for Mushroom Dataset (WITHOUT entropy display)

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# --------------------------------------------------
# Step 1: Load Mushroom Dataset
# --------------------------------------------------
df = pd.read_csv("mushrooms.csv")

print("First 5 rows of dataset:")
print(df.head())

# --------------------------------------------------
# Step 2: Label Encoding (All columns are categorical)
# --------------------------------------------------
label_encoders = {}
for col in df.columns:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    label_encoders[col] = le

print("\nEncoded Dataset Sample:")
print(df.head())

# --------------------------------------------------
# Step 3: Split Features and Target
# --------------------------------------------------
X = df.drop('class', axis=1)   # Features
y = df['class']                # Target (e=edible, p=poisonous)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0
)

# --------------------------------------------------
# Step 4: Train Decision Tree (Entropy)
# --------------------------------------------------
model = DecisionTreeClassifier(
    criterion='entropy',
    max_depth=5,
    random_state=0
)
model.fit(X_train, y_train)

print("\nDecision Tree Model Trained Successfully")

# --------------------------------------------------
# Step 5: Model Accuracy
# --------------------------------------------------
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print("Model Accuracy:", round(accuracy * 100, 2), "%")

# --------------------------------------------------
# Step 6: Classify New Mushroom Sample
# Using first row as example
# --------------------------------------------------
new_sample = pd.DataFrame({
    col: [X.iloc[0][col]] for col in X.columns
})

prediction = model.predict(new_sample)
result = label_encoders['class'].inverse_transform(prediction)

print("\nNew Mushroom Prediction:",
      "Edible" if result[0] == 'e' else "Poisonous")

# --------------------------------------------------
# Step 7: Plot Decision Tree
# --------------------------------------------------
plt.figure(figsize=(20, 10))
plot_tree(
    model,
    feature_names=X.columns,
    class_names=['Edible', 'Poisonous'],
    filled=True,
    rounded=True,
    impurity=False
)

plt.title("Decision Tree for Mushroom Classification")
plt.show()

# --------------------------------------------------
# Step 6: Predict a Custom Mushroom Sample
# --------------------------------------------------

sample_dict = {
    'cap-shape': 'x',
    'cap-surface': 's',
    'cap-color': 'n',
    'bruises': 't',
    'odor': 'a',
    'gill-attachment': 'f',
    'gill-spacing': 'c',
    'gill-size': 'b',
    'gill-color': 'k',
    'stalk-shape': 'e',
    'stalk-root': 'e',
    'stalk-surface-above-ring': 's',
    'stalk-surface-below-ring': 's',
    'stalk-color-above-ring': 'w',
    'stalk-color-below-ring': 'w',
    'veil-type': 'p',
    'veil-color': 'w',
    'ring-number': 'o',
    'ring-type': 'p',
    'spore-print-color': 'n',
    'population': 's',
    'habitat': 'g'
}

# Convert to DataFrame
new_mushroom = pd.DataFrame([sample_dict])

# Encode using previously fitted encoders
for col in new_mushroom.columns:
    new_mushroom[col] = label_encoders[col].transform(new_mushroom[col])

prediction = model.predict(new_mushroom)
result = label_encoders['class'].inverse_transform(prediction)

print("\nCustom Mushroom Prediction:",
      "Edible" if result[0] == 'e' else "Poisonous")
```
