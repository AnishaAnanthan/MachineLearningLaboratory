```python
from transformers import pipeline
from sklearn.metrics import accuracy_score, classification_report

# Step 1: Load pretrained sentiment analysis pipeline
sentiment_pipeline = pipeline("sentiment-analysis")

# Step 2: Dataset (same as Task 1)
texts = [
    "I love this product",
    "Very bad experience",
    "Amazing quality",
    "Not worth the money",
    "Excellent performance",
    "Terrible support",
    "Happy with purchase",
    "Waste of time",
    "Good value",
    "Disappointed"
]

true_labels = [
    "Positive", "Negative", "Positive", "Negative", "Positive",
    "Negative", "Positive", "Negative", "Positive", "Negative"
]

# Convert true labels to numeric
y_true = [1 if label == "Positive" else 0 for label in true_labels]

# Step 3: Get predictions from pretrained model
predictions = sentiment_pipeline(texts)

# Step 4: Convert predictions to numeric labels
y_pred = []
for pred in predictions:
    if pred['label'] == 'POSITIVE':
        y_pred.append(1)
    else:
        y_pred.append(0)

# Step 5: Evaluate performance
print("Accuracy:", accuracy_score(y_true, y_pred))
print("\nClassification Report:\n", classification_report(y_true, y_pred))

# Step 6: Test with a new sentence
test_sentence = "Excellent product"
result = sentiment_pipeline(test_sentence)

print("\nTest Sentence:", test_sentence)
print("Predicted Sentiment:", result[0]['label'])
print("Confidence Score:", result[0]['score'])
```
